# Token Counter & Cost Estimator

Estimate the token count and API cost of a prompt **before** sending it.

**Learning outcomes:**
- Understand what a token is
- Use `tiktoken` to count tokens for different models
- Compute cost from token count × price

```python
!pip install -q tiktoken
import tiktoken
```

```python
# Different models use different encodings
for model in ["gpt-4o", "gpt-3.5-turbo", "text-embedding-ada-002"]:
    enc = tiktoken.encoding_for_model(model)
    text = "The quick brown fox jumps over the lazy dog. " * 100
    print(f"{model:30s} -> {len(enc.encode(text)):6d} tokens")
```

```python
# Price table (USD per 1M tokens — update with current rates)
PRICING = {
    "gpt-4o":        {"input": 2.50,  "output": 10.00},
    "gpt-4o-mini":   {"input": 0.15,  "output": 0.60},
    "gpt-3.5-turbo": {"input": 0.50,  "output": 1.50},
}

def estimate_cost(model: str, prompt: str, expected_output_tokens: int = 500) -> float:
    enc = tiktoken.encoding_for_model(model)
    in_tokens = len(enc.encode(prompt))
    p = PRICING[model]
    cost = (in_tokens * p["input"] + expected_output_tokens * p["output"]) / 1_000_000
    return {"input_tokens": in_tokens, "output_tokens": expected_output_tokens, "cost_usd": cost}

estimate_cost("gpt-4o", "Summarize the attached 50-page contract:", 2000)
```
