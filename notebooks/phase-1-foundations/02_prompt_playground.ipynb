# Prompt Playground — Temperature & Sampling

Compare outputs across temperature / top-p / models on the same prompt.

**Learning outcomes:**
- See how temperature changes creativity vs determinism
- Understand top-p (nucleus) sampling
- Compare model behaviors

> No OpenAI key? Set `base_url="http://localhost:11434/v1"` and any model
> you pulled in Ollama (e.g. `qwen2.5:7b`).

```python
!pip install -q openai python-dotenv
import os
from openai import OpenAI
from dotenv import load_dotenv
load_dotenv()

# Use OpenAI, or Ollama by setting base_url + a dummy key
client = OpenAI(api_key=os.getenv("OPENAI_API_KEY", "ollama"),
                base_url=os.getenv("OPENAI_BASE_URL", None))
MODEL = os.getenv("PLAYGROUND_MODEL", "gpt-4o-mini")
```

```python
PROMPT = "Write a one-sentence tagline for a coffee shop run by cats."

for temp in [0.0, 0.7, 1.5]:
    r = client.chat.completions.create(
        model=MODEL,
        messages=[{"role": "user", "content": PROMPT}],
        temperature=temp,
        n=3,
    )
    print(f"\n=== temperature={temp} ===")
    for c in r.choices:
        print(f"  - {c.message.content.strip()}")
```
